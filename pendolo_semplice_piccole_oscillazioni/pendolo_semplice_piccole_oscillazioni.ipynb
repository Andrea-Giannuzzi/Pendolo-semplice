# Pendolo Semplice - Piccole Oscillazioni (Soluzione Analitica)

## Introduzione

Questo notebook simula il movimento di un **pendolo semplice** nel regime di **piccole oscillazioni** usando la **soluzione analitica chiusa**.

### Approssimazione

Per piccoli angoli ($\theta \ll 1$ rad), vale l'approssimazione:

$$\sin(\theta) \approx \theta$$

Quindi l'equazione non lineare:

$$\frac{d^2\theta}{dt^2} + \frac{g}{L} \sin(\theta) = 0$$

diventa:

$$\frac{d^2\theta}{dt^2} + \frac{g}{L} \theta = 0$$

Questa è l'**equazione di un oscillatore armonico semplice**, la cui soluzione analitica è:

$$\theta(t) = \theta_0 \cos(\omega t) + \frac{\omega_0}{\omega} \sin(\omega t)$$

dove $\omega = \sqrt{g/L}$ è la **frequenza angolare naturale** del pendolo.

```python
# Importare le librerie necessarie
import numpy as np
import matplotlib.pyplot as plt

# Configurazione di matplotlib per il notebook
%matplotlib inline
plt.style.use('seaborn-v0_8-darkgrid')
```

## Parametri Fisici

```python
# Parametri fisici
L = 1.0      # Lunghezza del pendolo (m)
g = 9.81     # Accelerazione di gravità (m/s²)

# Frequenza angolare naturale
omega = np.sqrt(g / L)
T = 2 * np.pi / omega  # Periodo dell'oscillazione

# Condizioni iniziali
theta0 = 0.1     # Angolo iniziale (rad) - piccolo per validare l'approssimazione
omega0 = 0.0     # Velocità angolare iniziale (rad/s)

# Parametri di simulazione
t_max = 3 * T    # Simulare 3 periodi completi
n_points = 1000  # Numero di punti temporali

print(f"Parametri della simulazione:")
print(f"  Lunghezza pendolo: L = {L} m")
print(f"  Gravità: g = {g} m/s²")
print(f"  Frequenza angolare: ω = √(g/L) = {omega:.4f} rad/s")
print(f"  Periodo: T = 2π/ω = {T:.4f} s")
print(f"  Numero di periodi simulati: {t_max/T:.1f}")
print(f"  Condizione iniziale: θ₀ = {theta0} rad ({np.degrees(theta0):.2f}°)")
print(f"  Velocità iniziale: ω₀ = {omega0} rad/s")
```

## Soluzioni Analitiche Chiuse

Definiamo le tre funzioni che descrivono il moto del pendolo in regime di piccole oscillazioni:

$$\theta(t) = \theta_0 \cos(\omega t) + \frac{\omega_0}{\omega} \sin(\omega t)$$

$$\omega(t) = \frac{d\theta}{dt} = -\theta_0 \omega \sin(\omega t) + \omega_0 \cos(\omega t)$$

$$\alpha(t) = \frac{d\omega}{dt} = -\frac{g}{L} \theta(t)$$

```python
def theta_analitica(t):
    """Soluzione analitica della posizione angolare θ(t)."""
    return theta0 * np.cos(omega * t) + (omega0 / omega) * np.sin(omega * t)

def velocita_angolare(t):
    """Soluzione analitica della velocità angolare ω(t) = dθ/dt."""
    return -theta0 * omega * np.sin(omega * t) + omega0 * np.cos(omega * t)

def accelerazione_angolare(t):
    """Soluzione analitica dell'accelerazione angolare α(t) = dω/dt."""
    return -(g / L) * theta_analitica(t)

# Vettore dei tempi
t_eval = np.linspace(0.0, t_max, n_points)

# Calcolare le tre quantità
theta = theta_analitica(t_eval)
omega_t = velocita_angolare(t_eval)
alpha = accelerazione_angolare(t_eval)

print(f"Soluzione analitica calcolata:")
print(f"  θ(t) varia tra {theta.min():.4f} e {theta.max():.4f} rad")
print(f"  ω(t) varia tra {omega_t.min():.4f} e {omega_t.max():.4f} rad/s")
print(f"  α(t) varia tra {alpha.min():.4f} e {alpha.max():.4f} rad/s²")
```

## Visualizzazione dei Risultati

Creiamo tre grafici che mostrano il moto del pendolo nel regime di piccole oscillazioni con soluzione analitica.

```python
# Creare tre subplot per i tre grafici
fig, axes = plt.subplots(3, 1, figsize=(12, 10))

# Grafico 1: Posizione angolare
axes[0].plot(t_eval, theta, 'steelblue', linewidth=2.5, label='θ(t) [soluzione analitica]')
axes[0].axhline(y=theta0, color='gray', linestyle='--', alpha=0.5, label=f'θ₀ = {theta0} rad')
axes[0].axhline(y=-theta0, color='gray', linestyle='--', alpha=0.5)
axes[0].set_xlabel('Tempo t (s)', fontsize=11)
axes[0].set_ylabel('Posizione angolare θ (rad)', fontsize=11)
axes[0].set_title('Piccole Oscillazioni: Posizione Angolare', fontsize=12, fontweight='bold')
axes[0].grid(True, alpha=0.3)
axes[0].legend(fontsize=10)

# Grafico 2: Velocità angolare
axes[1].plot(t_eval, omega_t, 'darkorange', linewidth=2.5, label='ω(t) [soluzione analitica]')
axes[1].set_xlabel('Tempo t (s)', fontsize=11)
axes[1].set_ylabel('Velocità angolare ω (rad/s)', fontsize=11)
axes[1].set_title('Piccole Oscillazioni: Velocità Angolare', fontsize=12, fontweight='bold')
axes[1].grid(True, alpha=0.3)
axes[1].legend(fontsize=10)

# Grafico 3: Accelerazione angolare
axes[2].plot(t_eval, alpha, 'crimson', linewidth=2.5, label='α(t) [soluzione analitica]')
axes[2].set_xlabel('Tempo t (s)', fontsize=11)
axes[2].set_ylabel('Accelerazione angolare α (rad/s²)', fontsize=11)
axes[2].set_title('Piccole Oscillazioni: Accelerazione Angolare', fontsize=12, fontweight='bold')
axes[2].grid(True, alpha=0.3)
axes[2].legend(fontsize=10)

plt.tight_layout()
plt.show()

print("✓ Grafici completati!")
```
